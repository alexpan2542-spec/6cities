```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    classification_report
)

# =====================================
# PATHS
# =====================================

ORIGINAL_FILE = (
    "../data2/Wuhan/01_original/"
    "Wuhan_WC_Samples_15000.csv"
)

CORRECTED_FILE = (
    "../data2/Wuhan/05_corrected/"
    "Wuhan_Corrected_15000.csv"
)

# =====================================
# FEATURES
# =====================================

features = [
    "B2",
    "B3",
    "B4",
    "B8",
    "B11",
    "B12",
    "NDVI",
    "NDBI",
    "MNDWI"
]

# =====================================
# RF FUNCTION
# =====================================

def run_rf(filepath, name):

    print("\n" + "=" * 50)
    print(name)
    print("=" * 50)

    df = pd.read_csv(filepath)

    X = df[features]
    y = df["Class"]

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.3,
        random_state=42,
        stratify=y
    )

    rf = RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    )

    rf.fit(X_train, y_train)

    pred = rf.predict(X_test)

    oa = accuracy_score(
        y_test,
        pred
    )

    print(f"OA = {oa:.6f}")

    print("\nConfusion Matrix")
    print(
        confusion_matrix(
            y_test,
            pred
        )
    )

    print("\nClassification Report")
    print(
        classification_report(
            y_test,
            pred,
            digits=4
        )
    )

    return oa

# =====================================
# RUN
# =====================================

oa_original = run_rf(
    ORIGINAL_FILE,
    "ORIGINAL + PIXEL"
)

oa_corrected = run_rf(
    CORRECTED_FILE,
    "CORRECTED + PIXEL"
)

# =====================================
# OA GAIN
# =====================================

gain = oa_corrected - oa_original

print("\n" + "=" * 50)
print("FINAL RESULT")
print("=" * 50)

print(
    f"Original OA   = {oa_original:.6f}"
)

print(
    f"Corrected OA  = {oa_corrected:.6f}"
)

print(
    f"OA Gain       = {gain:.6f}"
)

print(
    f"OA Gain (%)   = {gain*100:.2f}%"
)
```

```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# =================================================
# FILES
# =================================================

FEATURE_FILE = "../data2/Wuhan/06_3x3/Wuhan_3x3_Features.csv"

MANUAL_FILE = (
    "../data2/Wuhan/04_manual/"
    "Wuhan_Top150_Manual.csv"
)

# =================================================
# FEATURES
# =================================================

features = [

    "B2_mean", "B2_stdDev",

    "B3_mean", "B3_stdDev",

    "B4_mean", "B4_stdDev",

    "B8_mean", "B8_stdDev",

    "B11_mean", "B11_stdDev",

    "B12_mean", "B12_stdDev",

    "NDVI_mean", "NDVI_stdDev",

    "NDBI_mean", "NDBI_stdDev",

    "MNDWI_mean", "MNDWI_stdDev"
]

# =================================================
# RF FUNCTION
# =================================================

def run_rf(df, title):

    X = df[features]
    y = df["Class"]

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.3,
        random_state=42,
        stratify=y
    )

    rf = RandomForestClassifier(
        n_estimators=300,
        random_state=42,
        n_jobs=-1
    )

    rf.fit(X_train, y_train)

    pred = rf.predict(X_test)

    oa = accuracy_score(
        y_test,
        pred
    )

    print("\n" + "=" * 50)
    print(title)
    print("=" * 50)
    print(f"OA = {oa:.6f}")

    return oa

# =================================================
# LOAD ORIGINAL 3x3
# =================================================

df_original = pd.read_csv(
    FEATURE_FILE
)

# =================================================
# ORIGINAL + 3x3
# =================================================

oa_original = run_rf(
    df_original.copy(),
    "ORIGINAL + 3x3"
)

# =================================================
# LOAD MANUAL LABELS
# =================================================

manual = pd.read_csv(
    MANUAL_FILE
)

# =================================================
# APPLY CORRECTIONS
# =================================================

df_corrected = df_original.copy()

changes = 0

for _, row in manual.iterrows():

    oid = int(row["Original_ID"])

    human = int(row["Human_Class"])

    mask = (
        df_corrected["Original_ID"]
        == oid
    )

    if mask.sum() == 0:

        print(
            f"ID NOT FOUND: {oid}"
        )

        continue

    old = df_corrected.loc[
        mask,
        "Class"
    ].iloc[0]

    if old != human:

        df_corrected.loc[
            mask,
            "Class"
        ] = human

        changes += 1

print(
    f"\nLabels changed: {changes}"
)

# =================================================
# SAVE CORRECTED FILE
# =================================================

df_corrected.to_csv(
    "Wuhan_3x3_Features_Corrected.csv",
    index=False
)

# =================================================
# CORRECTED + 3x3
# =================================================

oa_corrected = run_rf(
    df_corrected,
    "CORRECTED + 3x3"
)

# =================================================
# FINAL RESULT
# =================================================

gain = oa_corrected - oa_original

print("\n" + "=" * 50)
print("FINAL RESULT")
print("=" * 50)

print(
    f"Original + 3x3   = {oa_original:.6f}"
)

print(
    f"Corrected + 3x3  = {oa_corrected:.6f}"
)

print(
    f"OA Gain          = {gain:.6f}"
)

print(
    f"OA Gain (%)      = {gain * 100:.2f}%"
)
```

```python
import pandas as pd
import numpy as np

# ==========================================
# FILES
# ==========================================

margin_df = pd.read_csv(
    "../data2/Wuhan/02_margin/Wuhan_MarginScores.csv"
)

f3 = pd.read_csv(
    "../data2/Wuhan/06_3x3/Wuhan_3x3_Features.csv"
)

# ==========================================
# MERGE
# ==========================================

df = margin_df.merge(
    f3[
        [
            "Original_ID",

            "B2_stdDev",
            "B3_stdDev",
            "B4_stdDev",
            "B8_stdDev"
        ]
    ],
    on="Original_ID",
    how="inner"
)

print("Rows:", len(df))

# ==========================================
# Boundary Score
# ==========================================

df["BoundaryScore"] = (

    df["B2_stdDev"] +

    df["B3_stdDev"] +

    df["B4_stdDev"] +

    df["B8_stdDev"]

)

# ==========================================
# Normalize
# ==========================================

df["MarginScore"] = 1 - df["margin"]

df["MarginScore"] = (
    (
        df["MarginScore"]
        -
        df["MarginScore"].min()
    )
    /
    (
        df["MarginScore"].max()
        -
        df["MarginScore"].min()
    )
)

df["BoundaryScore"] = (
    (
        df["BoundaryScore"]
        -
        df["BoundaryScore"].min()
    )
    /
    (
        df["BoundaryScore"].max()
        -
        df["BoundaryScore"].min()
    )
)

# ==========================================
# BAMS Score
# ==========================================

df["BAMS_Score"] = (

    0.7 * df["MarginScore"]

    +

    0.3 * df["BoundaryScore"]

)

# ==========================================
# Top150
# ==========================================

bams150 = (
    df
    .sort_values(
        "BAMS_Score",
        ascending=False
    )
    .head(150)
    .copy()
)

bams150["PointID"] = range(
    1,
    151
)

# ==========================================
# Save
# ==========================================

outfile = (
    "../data2/Wuhan/03_top150/"
    "Wuhan_BAMS150.csv"
)

bams150.to_csv(
    outfile,
    index=False
)

print("\nSaved:")
print(outfile)

print("\nTop20")

print(
    bams150[
        [
            "PointID",
            "Original_ID",
            "margin",
            "BoundaryScore",
            "BAMS_Score"
        ]
    ]
    .head(20)
)
```

```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
from sklearn.model_selection import train_test_split

# =====================================
# FILES
# =====================================

ORIGINAL_FILE = (
    "../data2/Nanjing/01_original/"
    "Nanjing_WC_Samples_15000.csv"
)

MANUAL_FILE = (
    "../data2/Nanjing/04_manual/"
    "Nanjing_BAMS150_Manual.csv"
)

# =====================================
# LOAD
# =====================================

orig = pd.read_csv(ORIGINAL_FILE)

manual = pd.read_csv(MANUAL_FILE)

manual["Human_Class"] = pd.to_numeric(
    manual["Human_Class"],
    errors="coerce"
)

manual = manual[
    manual["Human_Class"].notna()
]

# =====================================
# CREATE ERROR LABEL
# =====================================

merged = manual.merge(
    orig,
    on="Original_ID"
)

merged["Error_Label"] = (
    merged["Class"]
    !=
    merged["Human_Class"]
).astype(int)

print("\nError Statistics")

print(
    merged["Error_Label"]
    .value_counts()
)

# =====================================
# FEATURES
# =====================================

features = [
    "B2",
    "B3",
    "B4",
    "B8",
    "B11",
    "B12",
    "NDVI",
    "NDBI",
    "MNDWI"
]

X = merged[features]

y = merged["Error_Label"]

# =====================================
# TRAIN
# =====================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

rf = RandomForestClassifier(
    n_estimators=500,
    random_state=42,
    n_jobs=-1
)

rf.fit(
    X_train,
    y_train
)

pred = rf.predict(X_test)

print("\nClassification Report")

print(
    classification_report(
        y_test,
        pred,
        digits=4
    )
)

# =====================================
# FEATURE IMPORTANCE
# =====================================

imp = pd.DataFrame({
    "Feature": features,
    "Importance": rf.feature_importances_
})

imp = imp.sort_values(
    "Importance",
    ascending=False
)

print("\nTop Features")

print(imp)
```

```python
import pandas as pd

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

ORIGINAL_FILE = (
    "../data2/Changsha/01_original/"
    "Changsha_WC_Samples_15000.csv"
)

MANUAL_FILE = (
    "../data2/Changsha/04_manual/"
    "Changsha_BAMS150_Manual.csv"
)

CORRECTED_FILE = (
    "../data2/Changsha/05_corrected/"
    "Changsha_BAMS_Corrected_15000.csv"
)

df = pd.read_csv(ORIGINAL_FILE)
manual = pd.read_csv(MANUAL_FILE)

changes = 0

for _, row in manual.iterrows():

    if pd.isna(row["Human_Class"]):
        continue

    oid = int(row["Original_ID"])
    human = int(row["Human_Class"])

    mask = df["Original_ID"] == oid

    old_class = df.loc[
        mask,
        "Class"
    ].iloc[0]

    if old_class != human:

        df.loc[
            mask,
            "Class"
        ] = human

        changes += 1

df.to_csv(
    CORRECTED_FILE,
    index=False
)

print("Labels changed:", changes)

features = [
    "B2","B3","B4","B8",
    "B11","B12",
    "NDVI","NDBI","MNDWI"
]

X = df[features]
y = df["Class"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

rf = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

pred = rf.predict(X_test)

oa = accuracy_score(y_test, pred)

print("\n==============================")
print("CHANGSHA BAMS + PIXEL")
print("==============================")
print(f"OA = {oa:.6f}")
```
