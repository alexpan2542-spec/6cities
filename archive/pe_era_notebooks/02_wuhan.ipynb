```python
import pandas as pd

df = pd.read_csv(
    "../data/Wuhan_WC_Samples_15000.csv"
)

df.head()
```

```python
df["Class"].value_counts()
```

```python
features = [
    "B2",
    "B3",
    "B4",
    "B8",
    "B11",
    "B12",
    "NDVI",
    "NDBI",
    "MNDWI"
]

X = df[features]
y = df["Class"]
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)
```

```python
import numpy as np
prob = rf.predict_proba(X_test)

sorted_prob = np.sort(
    prob,
    axis=1
)

margin = (
    sorted_prob[:, -1]
    -
    sorted_prob[:, -2]
)

hard_idx = np.argsort(margin)
```

```python
hard_samples = df.loc[
    X_test.iloc[hard_idx[:100]].index,
    ['lon','lat','Class']
].copy()

hard_samples['margin'] = (
    margin[hard_idx[:100]]
)

hard_samples.to_csv(
    "../data/Wuhan_top100.csv",
    index=False
)

hard_samples.head()
```
